# Points
### nyc farmers markets colored by how clusteres they are

```python
# NYC Farmers Markets
markets_url = (
    "https://data.cityofnewyork.us/api/v3/views/"
    "8vwk-6iz2/export.csv?accessType=DOWNLOAD"
)

markets_df = pd.read_csv(markets_url)

# Inspect the columns
print(markets_df.columns.tolist())
print(markets_df.head())
```

```python
# NYC Borough Boundaries
boroughs_url = (
    "https://data.cityofnewyork.us/api/v3/views/"
    "gthc-hcne/query.geojson?accessType=DOWNLOAD"
)

boroughs = gpd.read_file(boroughs_url)

print(boroughs.head())
print(boroughs.crs)
```

```python

# Convert coordinates and year to numeric
markets_df["Latitude"] = pd.to_numeric(
    markets_df["Latitude"], errors="coerce"
)

markets_df["Longitude"] = pd.to_numeric(
    markets_df["Longitude"], errors="coerce"
)

markets_df["Year"] = pd.to_numeric(
    markets_df["Year"], errors="coerce"
)

# Remove rows with missing coordinates
markets_df = markets_df.dropna(
    subset=["Latitude", "Longitude"]
).copy()

# Keep the most recent year available
latest_year = int(markets_df["Year"].max())

markets_latest = markets_df[
    markets_df["Year"] == latest_year
].copy()

# Remove duplicate market locations
markets_latest = markets_latest.drop_duplicates(
    subset=["Latitude", "Longitude"]
).copy()

print("Year selected:", latest_year)
print("Number of market locations:", len(markets_latest))
```

```python
markets = gpd.GeoDataFrame(
    markets_latest,
    geometry=gpd.points_from_xy(
        markets_latest["Longitude"],
        markets_latest["Latitude"]
    ),
    crs="EPSG:4326"
)

markets.head()
```

```python
# Project both layers into NYC's projected coordinate system
markets_proj = markets.to_crs(epsg=2263)
boroughs_proj = boroughs.to_crs(epsg=2263)

# Extract point coordinates in projected units (US survey feet)
coords = np.column_stack([
    markets_proj.geometry.x,
    markets_proj.geometry.y
])

# Find the two nearest points:
# 1. The market itself
# 2. Its nearest other market
nn = NearestNeighbors(n_neighbors=2)
nn.fit(coords)

distances, indices = nn.kneighbors(coords)

# The first distance is 0 because it is the market itself.
# The second distance is to the nearest other market.
markets_proj["nearest_market_m"] = distances[:, 1] * 0.3048006096

# Optional: distance to the three nearest OTHER markets
nn3 = NearestNeighbors(n_neighbors=4)
nn3.fit(coords)

distances_3, indices_3 = nn3.kneighbors(coords)

markets_proj["mean_3_nearest_m"] = (
    distances_3[:, 1:4].mean(axis=1) * 0.3048006096
)

markets_proj[
    ["Market Name", "nearest_market_m", "mean_3_nearest_m"]
].head()
```

```python

fig, ax = plt.subplots(figsize=(12, 12))

# Borough boundaries underneath the points
boroughs_proj.plot(
    ax=ax,
    color="whitesmoke",
    edgecolor="black",
    linewidth=1
)

# Farmers markets colored by proximity
markets_proj.plot(
    ax=ax,
    column="mean_3_nearest_m",
    cmap="viridis",
    markersize=65,
    alpha=0.95,
    edgecolor="black",
    linewidth=0.4,
    legend=True,
    legend_kwds={
        "label": "Mean distance to 3 nearest markets (meters)",
        "shrink": 0.65
    }
)

# Zoom to the borough boundaries, with a small margin
xmin, ymin, xmax, ymax = boroughs_proj.total_bounds

x_margin = (xmax - xmin) * 0.03
y_margin = (ymax - ymin) * 0.03

ax.set_xlim(xmin - x_margin, xmax + x_margin)
ax.set_ylim(ymin - y_margin, ymax + y_margin)

ax.set_title(
    "NYC Farmers Markets: Spatial Proximity",
    fontsize=16
)

ax.set_axis_off()
plt.tight_layout()
plt.show()
```

```python
# Load Recognized Shop Healthy Stores
stores_url = (
    "https://data.cityofnewyork.us/api/v3/views/"
    "ud4g-9x9z/export.csv?accessType=DOWNLOAD"
)

stores_df = pd.read_csv(stores_url)

# Inspect the data
print(stores_df.columns.tolist())
print(stores_df.head())
```

```python

# Convert coordinates to numeric values
stores_df["Latitude"] = pd.to_numeric(
    stores_df["Latitude"], errors="coerce"
)

stores_df["Longitude"] = pd.to_numeric(
    stores_df["Longitude"], errors="coerce"
)

# Remove records without coordinates
stores_df = stores_df.dropna(
    subset=["Latitude", "Longitude"]
).copy()

# Convert to a GeoDataFrame
stores = gpd.GeoDataFrame(
    stores_df,
    geometry=gpd.points_from_xy(
        stores_df["Longitude"],
        stores_df["Latitude"]
    ),
    crs="EPSG:4326"
)

# Project to the same coordinate system as your existing map
stores_proj = stores.to_crs(epsg=2263)

print("Number of Shop Healthy stores:", len(stores_proj))
```

```python

fig, ax = plt.subplots(figsize=(12, 12))

# 1. Borough boundaries
boroughs_proj.plot(
    ax=ax,
    color="whitesmoke",
    edgecolor="black",
    linewidth=0.8,
    zorder=1
)

# 2. Farmers markets: colored by proximity
markets_proj.plot(
    ax=ax,
    column="mean_3_nearest_m",
    cmap="viridis",
    markersize=55,
    edgecolor="black",
    linewidth=0.3,
    alpha=0.9,
    legend=True,
    legend_kwds={
        "label": "Mean distance to 3 nearest farmers markets (meters)",
        "shrink": 0.65
    },
    zorder=2
)

# 3. Recognized Shop Healthy stores: red triangles
stores_proj.plot(
    ax=ax,
    color="red",
    marker="^",
    markersize=22,
    edgecolor="black",
    linewidth=0.3,
    alpha=0.8,
    zorder=3
)

# Zoom to NYC borough boundaries
xmin, ymin, xmax, ymax = boroughs_proj.total_bounds

x_margin = (xmax - xmin) * 0.03
y_margin = (ymax - ymin) * 0.03

ax.set_xlim(xmin - x_margin, xmax + x_margin)
ax.set_ylim(ymin - y_margin, ymax + y_margin)

# Title and legend
ax.set_title(
    "NYC Farmers Markets and Recognized Shop Healthy Stores",
    fontsize=15,
    pad=15
)

from matplotlib.lines import Line2D

legend_items = [
    Line2D(
        [0], [0],
        marker="^",
        color="w",
        markerfacecolor="red",
        markeredgecolor="black",
        markersize=8,
        label="Recognized Shop Healthy stores"
    )
]

ax.legend(
    handles=legend_items,
    loc="lower left",
    frameon=True
)

ax.set_axis_off()
plt.tight_layout()
plt.show()
```

```python
dem_path = Path("data/dem/USGS_13_n46w122_20260202.tif")
output_path = Path("outputs/mount_hood_dem_clipped.tif")

west = -121.825
south = 45.275
east = -121.575
north = 45.45

with rasterio.open(dem_path) as src:
    window = from_bounds(
        west, south, east, north,
        transform=src.transform
    )

    dem = src.read(1, window=window, masked=True)
    transform = src.window_transform(window)

    profile = src.profile.copy()
    profile.update(
        height=dem.shape[0],
        width=dem.shape[1],
        transform=transform
    )

    output_path.parent.mkdir(parents=True, exist_ok=True)

    nodata = src.nodata if src.nodata is not None else -9999.0
    profile.update(nodata=nodata)

    with rasterio.open(output_path, "w", **profile) as dst:
        dst.write(dem.filled(nodata), 1)

print(f"Saved cropped DEM to: {output_path}")
print(f"Crop dimensions: {dem.shape}")

# Preview the crop
plt.figure(figsize=(10, 10))
plt.imshow(
    dem,
    extent=[west, east, south, north],
    origin="upper",
    cmap="terrain"
)
plt.title("Mount Hood Zoomed DEM")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```

```python
with rasterio.open(dem_path) as src:
    dem_crs = src.crs
    bounds = src.bounds

dem_boundary = box(
    bounds.left,
    bounds.bottom,
    bounds.right,
    bounds.top
)

recreation_check = recreation.to_crs(dem_crs).copy()

recreation_check["dem_status"] = recreation_check.geometry.apply(
    lambda geom: (
        "Outside DEM" if not geom.intersects(dem_boundary)
        else "Fully inside DEM" if dem_boundary.covers(geom)
        else "Partially outside DEM"
    )
)

print(recreation_check["dem_status"].value_counts())
```

```python
# Keep only polygons fully within the DEM boundaries
recreation_clipped = recreation_check[
    recreation_check["dem_status"] == "Fully inside DEM"
].copy()

print("Original polygon count:", len(recreation_check))
print("Remaining polygon count:", len(recreation_clipped))
print("Removed polygon count:", len(recreation_check) - len(recreation_clipped))
```

```python
dem_path = "outputs/mount_hood_dem_clipped.tif"

# Read the DEM
with rasterio.open(dem_path) as src:
    dem = src.read(1, masked=True)
    dem_crs = src.crs
    extent = plotting_extent(src)

recreation_map = recreation_clipped.to_crs(dem_crs)

print("DEM CRS:", dem_crs)
print("Recreation polygons:", len(recreation_map))
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

im = ax.imshow(
    dem,
    extent=extent,
    origin="upper",
    cmap="terrain"
)

recreation_map.boundary.plot(
    ax=ax,
    color="blue",
    linewidth=1.2
)

cbar = fig.colorbar(im, ax=ax, shrink=0.7, pad=0.02)
cbar.set_label("Elevation (DEM units)")

ax.set_title("Mount Hood, Oregon: Elevation and Winter Recreation Areas",
             fontsize=15, pad=14)
ax.set_xlabel("Longitude / projected easting")
ax.set_ylabel("Latitude / projected northing")
ax.set_aspect("equal")

plt.tight_layout()
plt.show()
```

```python

with rasterio.open(dem_path) as src:
    dem_crs = src.crs
    dem_boundary = gpd.GeoDataFrame(
        geometry=[box(*src.bounds)],
        crs=dem_crs
    )

contours_map = contours.to_crs(dem_crs)

contours_clipped = gpd.clip(
    contours_map,
    dem_boundary
)

print("Original contours:", len(contours_map))
print("Clipped contours:", len(contours_clipped))
```

```python

import matplotlib.pyplot as plt
import numpy as np
import geopandas as gpd
import rasterio

from rasterio.warp import (
    calculate_default_transform,
    reproject,
    Resampling
)
from pyproj import CRS
from matplotlib.patches import Rectangle

# ==================================================
# 1. DEFINE AZIMUTHAL EQUIDISTANT PROJECTION
# ==================================================

# Center the projection on Mount Hood
target_crs = CRS.from_proj4(
    "+proj=aeqd "
    "+lat_0=45.3735 "
    "+lon_0=-121.6959 "
    "+datum=WGS84 "
    "+units=m "
    "+no_defs"
)

# ==================================================
# 2. READ AND REPROJECT THE DEM
# ==================================================

with rasterio.open(dem_path) as src:

    source_dem = src.read(1, masked=True).astype("float32").filled(np.nan)

    source_crs = src.crs
    source_transform = src.transform

    # Calculate projected raster dimensions and transform
    dst_transform, dst_width, dst_height = calculate_default_transform(
        source_crs,
        target_crs,
        src.width,
        src.height,
        *src.bounds
    )

    # Create destination raster
    dem_az = np.full(
        (dst_height, dst_width),
        np.nan,
        dtype="float32"
    )

    # Reproject elevation data
    reproject(
        source=source_dem,
        destination=dem_az,
        src_transform=source_transform,
        src_crs=source_crs,
        src_nodata=np.nan,
        dst_transform=dst_transform,
        dst_crs=target_crs,
        dst_nodata=np.nan,
        resampling=Resampling.bilinear
    )

# ==================================================
# 3. CORRECTLY CALCULATE THE PROJECTED DEM EXTENT
# ==================================================

left = dst_transform.c
top = dst_transform.f

right = left + dst_width * dst_transform.a
bottom = top + dst_height * dst_transform.e

# Matplotlib expects (left, right, bottom, top)
extent_az = (left, right, bottom, top)

print("Projected DEM shape:", dem_az.shape)
print("Valid DEM pixels:", np.isfinite(dem_az).sum())
print("Projected extent:", extent_az)

# ==================================================
# 4. REPROJECT CONTOURS AND RECREATION AREAS
# ==================================================

contours_az = contours_map.to_crs(target_crs).copy()
recreation_az = recreation_clipped.to_crs(target_crs).copy()

# Repair invalid geometries
contours_az["geometry"] = contours_az.geometry.make_valid()
recreation_az["geometry"] = recreation_az.geometry.make_valid()

# Remove missing and empty geometries
contours_az = contours_az[
    contours_az.geometry.notna()
    & ~contours_az.geometry.is_empty
].copy()

recreation_az = recreation_az[
    recreation_az.geometry.notna()
    & ~recreation_az.geometry.is_empty
].copy()

# ==================================================
# 5. HIGH-CONTRAST DEM DISPLAY SETTINGS
# ==================================================

valid_elevation = dem_az[np.isfinite(dem_az)]

vmin, vmax = np.percentile(
    valid_elevation,
    [2, 98]
)

# ==================================================
# 6. CREATE THE MAP
# ==================================================

fig, ax = plt.subplots(
    figsize=(12, 12),
    facecolor="black"
)

ax.set_facecolor("black")

# --------------------------------------------------
# A. Plot grayscale DEM
# --------------------------------------------------

ax.imshow(
    dem_az,
    extent=extent_az,
    origin="upper",
    cmap="gray_r",
    vmin=vmin,
    vmax=vmax,
    interpolation="nearest",
    zorder=1
)

# --------------------------------------------------
# B. Plot all contours in pale blue
# --------------------------------------------------

contours_az.plot(
    ax=ax,
    color="#97C1E6", #0165FC
    linewidth=0.35,
    alpha=0.8,
    zorder=2
)

# --------------------------------------------------
# C. Highlight contours inside recreation areas
# --------------------------------------------------

# Bright orange for recreation areas
bright_colors = ["#FF5F1F"]

for i, (_, area) in enumerate(recreation_az.iterrows()):

    # Create a GeoDataFrame for this individual area
    area_gdf = gpd.GeoDataFrame(
        geometry=[area.geometry],
        crs=recreation_az.crs
    )

    # Draw the recreation-area boundary
    area_gdf.boundary.plot(
        ax=ax,
        color=bright_colors[i % len(bright_colors)],
        linewidth=1.0,
        zorder=4
    )

    # Select contour lines intersecting this recreation area
    area_contours = gpd.clip(
        contours_az,
        area_gdf
    )

    # Plot the clipped contour segments
    if not area_contours.empty:
        area_contours.plot(
            ax=ax,
            color=bright_colors[i % len(bright_colors)],
            linewidth=0.65,
            alpha=1,
            zorder=3
        )

# --------------------------------------------------
# Plot OpenStreetMap roads in bright red
# --------------------------------------------------


# Bright red roads
roads_az.plot(
    ax=ax,
    color="#00008B",
    linewidth=0.65,
    alpha=0.4,
    zorder=2.1
)
# ==================================================
# 7. SET MAP EXTENT
# ==================================================

xmin, xmax, ymin, ymax = extent_az

xpad = (xmax - xmin) * 0.02
ypad = (ymax - ymin) * 0.02

ax.set_xlim(
    xmin - xpad,
    xmax + xpad
)

ax.set_ylim(
    ymin - ypad,
    ymax + ypad
)

# ==================================================
# 8. TITLE AND FINAL STYLING
# ==================================================

ax.set_title(
    "MOUNT HOOD, OREGON",
    color="white",
    fontsize=18,
    fontweight="bold",
    pad=16
)

ax.set_aspect("equal")
ax.set_axis_off()

plt.tight_layout()

# ==================================================
# 9. EXPORT AND DISPLAY
# ==================================================

# Optional: save a high-resolution raster image
# fig.savefig(
#     "mount_hood_azimuthal.png",
#     dpi=300,
#     bbox_inches="tight",
#     facecolor=fig.get_facecolor()
# )

# Optional: save a vector SVG
# fig.savefig(
#     "mount_hood_azimuthal.svg",
#     format="svg",
#     bbox_inches="tight",
#     facecolor=fig.get_facecolor()
# )

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import geopandas as gpd
import rasterio
import osmnx as ox

from rasterio.warp import (
    calculate_default_transform,
    reproject,
    Resampling
)
from pyproj import CRS
from shapely.ops import unary_union
from matplotlib.lines import Line2D

# ==================================================
# 1. DEFINE AZIMUTHAL EQUIDISTANT PROJECTION
# ==================================================

target_crs = CRS.from_proj4(
    "+proj=aeqd "
    "+lat_0=45.3735 "
    "+lon_0=-121.6959 "
    "+datum=WGS84 "
    "+units=m "
    "+no_defs"
)

# ==================================================
# 2. READ AND REPROJECT THE DEM
# ==================================================

with rasterio.open(dem_path) as src:

    source_dem = (
        src.read(1, masked=True)
        .astype("float32")
        .filled(np.nan)
    )

    source_crs = src.crs
    source_transform = src.transform

    dst_transform, dst_width, dst_height = calculate_default_transform(
        source_crs,
        target_crs,
        src.width,
        src.height,
        *src.bounds
    )

    dem_az = np.full(
        (dst_height, dst_width),
        np.nan,
        dtype="float32"
    )

    reproject(
        source=source_dem,
        destination=dem_az,
        src_transform=source_transform,
        src_crs=source_crs,
        src_nodata=np.nan,
        dst_transform=dst_transform,
        dst_crs=target_crs,
        dst_nodata=np.nan,
        resampling=Resampling.bilinear
    )

# ==================================================
# 3. CALCULATE THE CORRECT DEM EXTENT
# ==================================================

left = dst_transform.c
top = dst_transform.f

right = left + dst_width * dst_transform.a
bottom = top + dst_height * dst_transform.e

extent_az = (left, right, bottom, top)

# ==================================================
# 4. REPROJECT AND CLEAN VECTOR LAYERS
# ==================================================

contours_az = contours_map.to_crs(target_crs).copy()
recreation_az = recreation_clipped.to_crs(target_crs).copy()
roads_az = roads_az.to_crs(target_crs).copy()

# Repair invalid geometries
contours_az["geometry"] = contours_az.geometry.make_valid()
recreation_az["geometry"] = recreation_az.geometry.make_valid()
roads_az["geometry"] = roads_az.geometry.make_valid()

# Remove missing and empty geometries
contours_az = contours_az[
    contours_az.geometry.notna()
    & ~contours_az.geometry.is_empty
].copy()

recreation_az = recreation_az[
    recreation_az.geometry.notna()
    & ~recreation_az.geometry.is_empty
].copy()

roads_az = roads_az[
    roads_az.geometry.notna()
    & ~roads_az.geometry.is_empty
].copy()

# ==================================================
# 5. IDENTIFY RECREATION AREA NAMES
# ==================================================

print("Available recreation-area columns:")
print(recreation_az.columns.tolist())

candidate_cols = [
    "RECAREANAME",
    "SITENAME",
    "SiteName",
    "site_name",
    "SITE_NAME",
    "RecSiteName",
    "recreation_name",
    "NAME",
    "Name",
    "name",
    "UNIT_NAME",
    "UnitName"
]

name_col = next(
    (col for col in candidate_cols if col in recreation_az.columns),
    None
)

if name_col is not None:
    print(f"Using '{name_col}' for recreation-area labels.")
else:
    print("No name field found. Using numbered labels.")

area_labels = []

for i, (_, area) in enumerate(recreation_az.iterrows(), start=1):

    if name_col is not None:
        value = area[name_col]

        if (
            value is not None
            and not (
                isinstance(value, (float, np.floating))
                and np.isnan(value)
            )
            and str(value).strip()
        ):
            label = str(value)
        else:
            label = f"Recreation Area {i}"
    else:
        label = f"Recreation Area {i}"

    area_labels.append(label)

# ==================================================
# 6. ASSIGN DISTINCT LINE STYLES TO RECREATION AREAS
# ==================================================

from matplotlib.lines import Line2D

# Bright colors
boundary_color = "#0047AB"  # Bright blue outlines
contour_highlight_color = "#f18701"  # Bright orange contours

# Different boundary line styles
line_styles = [
    "-",                      # 1. Solid
    (0, (8, 2)),               # 2. Long dashes
    (0, (1, 1)),               # 3. Dense dots
    (0, (6, 2, 1, 2)),         # 4. Long dash, dot
    (0, (8, 2, 1, 2, 1, 2)),  # 5. Long dash, two dots
]

# Keep one legend entry per unique area name
unique_names = list(dict.fromkeys(area_labels))

style_by_name = {
    name: line_styles[i % len(line_styles)]
    for i, name in enumerate(unique_names)
}

# ==================================================
# 7. REMOVE ROADS INSIDE RECREATION AREAS
# ==================================================

# Combine all recreation polygons into one geometry.
recreation_union = unary_union(recreation_az.geometry)

# Subtract the recreation areas from each road geometry.
# This leaves road segments only where they are outside
# recreation areas.

roads_outside_recreation = roads_az.copy()

roads_outside_recreation["geometry"] = (
    roads_outside_recreation.geometry.difference(
        recreation_union
    )
)

# Remove empty and missing geometries
roads_outside_recreation = roads_outside_recreation[
    roads_outside_recreation.geometry.notna()
    & ~roads_outside_recreation.geometry.is_empty
].copy()

# Keep only line geometries
roads_outside_recreation = roads_outside_recreation[
    roads_outside_recreation.geometry.geom_type.isin(
        ["LineString", "MultiLineString"]
    )
].copy()

print(
    f"Road features outside recreation areas: "
    f"{len(roads_outside_recreation)}"
)

# ==================================================
# 8. SET DEM CONTRAST
# ==================================================

valid_elevation = dem_az[np.isfinite(dem_az)]

vmin, vmax = np.percentile(
    valid_elevation,
    [2, 98]
)

# ==================================================
# 9. CREATE THE MAP
# ==================================================

fig, ax = plt.subplots(
    figsize=(17, 12),
    facecolor="black"
)

ax.set_facecolor("black")

# --------------------------------------------------
# A. Plot grayscale DEM
# --------------------------------------------------

ax.imshow(
    dem_az,
    extent=extent_az,
    origin="upper",
    cmap="gray_r",
    vmin=vmin,
    vmax=vmax,
    interpolation="nearest",
    zorder=1
)

# # --------------------------------------------------
# # B. Plot roads outside recreation areas
# # --------------------------------------------------

# roads_outside_recreation.plot(
#     ax=ax,
#     color="#C4A484",
#     linewidth=0.65,
#     alpha=0.55,
#     zorder=2
# )

# --------------------------------------------------
# C. Plot all elevation contours in pale blue
# --------------------------------------------------

contours_az.plot(
    ax=ax,
    color="#97C1E6",
    linewidth=0.35,
    alpha=0.8,
    zorder=3
)

# ==================================================
# 10. HIGHLIGHT RECREATION AREAS
# ==================================================

legend_items = []

for label in unique_names:

    line_style = style_by_name[label]

    # Get all polygons associated with this recreation-area name
    area_indices = [
        i for i, area_label in enumerate(area_labels)
        if area_label == label
    ]

    area_parts = recreation_az.iloc[area_indices]

    # --------------------------------------------------
    # A. DRAW BRIGHT BLUE BOUNDARIES WITH DISTINCT STYLES
    # --------------------------------------------------

    area_parts.boundary.plot(
        ax=ax,
        color=boundary_color,
        linewidth=2.0,
        linestyle=line_style,
        alpha=1.0,
        zorder=5
    )

    # --------------------------------------------------
    # B. HIGHLIGHT CONTOURS INSIDE RECREATION AREAS IN ORANGE
    # --------------------------------------------------

    try:
        area_union = area_parts.geometry.union_all()

        area_contours = gpd.clip(
            contours_az,
            area_union
        )

        if not area_contours.empty:
            area_contours.plot(
                ax=ax,
                color=contour_highlight_color,
                linewidth=0.8,
                linestyle="-",  # All highlighted contours are solid
                alpha=1.0,
                zorder=4
            )

    except Exception as e:
        print(f"Could not clip contours for {label}: {e}")

    # --------------------------------------------------
    # C. CREATE LEGEND ENTRY FOR EACH RECREATION AREA
    # --------------------------------------------------

    legend_items.append(
        Line2D(
            [0],
            [0],
            color=boundary_color,
            linewidth=2.0,
            linestyle=line_style,
            label=label
        )
    )

# ==================================================
# 11. ADD LEGEND
# ==================================================

legend = ax.legend(
    handles=legend_items,
    title="WINTER RECREATION AREAS",
    loc="upper left",
    bbox_to_anchor=(1.01, 1.0),
    frameon=True,
    facecolor="#111111",
    edgecolor="#777777",
    fontsize=8,
    title_fontsize=10,
    labelspacing=1.0,
    borderpad=1.2,
    handlelength=3.5
)

legend.get_title().set_color("white")
legend.get_title().set_fontweight("bold")

for text in legend.get_texts():
    text.set_color("white")
# ==================================================
# 11. ADD LEGEND
# ==================================================

legend = ax.legend(
    handles=legend_items,
    title="WINTER RECREATION AREAS",
    loc="upper left",
    bbox_to_anchor=(1.01, 1.0),
    frameon=True,
    facecolor="#111111",
    edgecolor="#777777",
    fontsize=8,
    title_fontsize=10,
    labelspacing=1.0,
    borderpad=1.2,
    handlelength=3.5
)

legend.get_title().set_color("white")
legend.get_title().set_fontweight("bold")

for text in legend.get_texts():
    text.set_color("white")

# ==================================================
# 12. SET MAP EXTENT
# ==================================================

xmin, xmax, ymin, ymax = extent_az

xpad = (xmax - xmin) * 0.02
ypad = (ymax - ymin) * 0.02

ax.set_xlim(
    xmin - xpad,
    xmax + xpad
)

ax.set_ylim(
    ymin - ypad,
    ymax + ypad
)

# ==================================================
# 13. TITLE AND FINAL STYLING
# ==================================================

ax.set_title(
    "MOUNT HOOD, OREGON",
    color="white",
    fontsize=18,
    fontweight="bold",
    pad=16
)

ax.set_aspect("equal")
ax.set_axis_off()

# Leave room for the legend on the right
fig.subplots_adjust(
    left=0.02,
    right=0.75,
    top=0.92,
    bottom=0.03
)

# ==================================================
# 14. EXPORT AND DISPLAY
# ==================================================

# Optional: export a high-quality SVG for Illustrator.
# Uncomment these lines to save the map.

fig.savefig(
    "mount_hood_recreation_legend.svg",
    format="svg",
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)

plt.show()
```

```python
# Work with a copy of the original data
canopy_map = canopy.copy()

# Remove missing and empty geometries
canopy_map = canopy_map[
    canopy_map.geometry.notna()
    & ~canopy_map.geometry.is_empty
].copy()

# Remove features with missing or non-positive heights
canopy_map = canopy_map[
    canopy_map["avg_height"].notna()
    & (canopy_map["avg_height"] > 0)
].copy()

# Repair invalid geometries if necessary
invalid = ~canopy_map.geometry.is_valid

if invalid.any():
    canopy_map.loc[invalid, "geometry"] = (
        canopy_map.loc[invalid, "geometry"].make_valid()
    )

# Reproject from geographic coordinates to Pennsylvania South State Plane
canopy_map = canopy_map.to_crs("EPSG:2272")

# Calculate polygon areas in square feet
canopy_map["area_sqft"] = canopy_map.geometry.area

# Remove any geometries that remain unsuitable
canopy_map = canopy_map[
    canopy_map.geometry.notna()
    & ~canopy_map.geometry.is_empty
    & canopy_map.geometry.is_valid
    & (canopy_map["area_sqft"] > 0)
].copy()

print("Valid canopy polygons:", len(canopy_map))
print("\nArea distribution (square feet):")
print(canopy_map["area_sqft"].describe())
```

```python
# Choose how many of the largest polygons to display
N_LARGEST = 300

largest_canopies = (
    canopy_map
    .nlargest(N_LARGEST, "area_sqft")
    .copy()
    .reset_index(drop=True)
)

print(f"Selected {len(largest_canopies)} canopy polygons")

print("\nLargest canopy areas:")
print(
    largest_canopies[
        ["objectid", "avg_height", "area_sqft"]
    ].head(10)
)
```

```python
def polygon_parts(geom):
    """
    Yield individual Polygon components from a Polygon or MultiPolygon.
    """
    if geom.geom_type == "Polygon":
        yield geom

    elif geom.geom_type == "MultiPolygon":
        yield from geom.geoms

    elif geom.geom_type == "GeometryCollection":
        for part in geom.geoms:
            yield from polygon_parts(part)


def make_side_faces(ring, height):
    """
    Create vertical wall faces along a polygon ring.
    Coordinates are (x, y, z).
    """
    coords = list(ring.coords)
    faces = []

    for i in range(len(coords) - 1):
        x1, y1 = coords[i][:2]
        x2, y2 = coords[i + 1][:2]

        # Skip zero-length edges
        if x1 == x2 and y1 == y2:
            continue

        faces.append([
            (x1, y1, 0),
            (x2, y2, 0),
            (x2, y2, height),
            (x1, y1, height)
        ])

    return faces


def make_extrusion_faces(geom, height):
    """
    Create top triangles, bottom triangles, and vertical side walls
    for all Polygon components of a geometry.
    """
    faces = []

    for poly in polygon_parts(geom):
        if poly.is_empty or poly.area <= 0:
            continue

        # Triangulate the polygon and retain triangles lying inside it.
        # This helps preserve concave outlines and holes.
        triangles = triangulate(poly)

        for tri in triangles:
            if not poly.covers(tri):
                continue

            coords = list(tri.exterior.coords)[:3]

            top_face = [
                (x, y, height)
                for x, y in coords
            ]

            bottom_face = [
                (x, y, 0)
                for x, y in coords
            ]

            faces.append(top_face)
            faces.append(bottom_face)

        # Add exterior walls
        faces.extend(
            make_side_faces(poly.exterior, height)
        )

        # Add walls around any interior holes
        for interior in poly.interiors:
            faces.extend(
                make_side_faces(interior, height)
            )

    return faces


# Collect faces for the selected canopy polygons
all_faces = []
face_groups = []

for _, row in largest_canopies.iterrows():
    height = float(row["avg_height"])

    faces = make_extrusion_faces(
        row.geometry,
        height
    )

    all_faces.extend(faces)

    # Keep track of which faces belong to each canopy
    face_groups.extend([height] * len(faces))

print("Total 3D faces:", len(all_faces))
```

```python
# Create the figure
fig = plt.figure(figsize=(16, 16), facecolor="#F5F2EA")

ax = fig.add_subplot(111, projection="3d")
ax.set_facecolor("#F5F2EA")

# Draw all extrusion faces
canopy_collection = Poly3DCollection(
    all_faces,
    facecolors="#356B49",
    edgecolors="#244B35",
    linewidths=0.08,
    alpha=1.0,
    zsort="average"
)

ax.add_collection3d(canopy_collection)

# Determine the horizontal extent of the selected polygons
minx, miny, maxx, maxy = largest_canopies.total_bounds

# Center the view on the selected canopy patches
center_x = (minx + maxx) / 2
center_y = (miny + maxy) / 2

width = maxx - minx
height_xy = maxy - miny

# Give the map some breathing room
padding = 0.06

ax.set_xlim(
    center_x - width * (0.5 + padding),
    center_x + width * (0.5 + padding)
)

ax.set_ylim(
    center_y - height_xy * (0.5 + padding),
    center_y + height_xy * (0.5 + padding)
)

# Vertical extent
max_height = largest_canopies["avg_height"].max()
ax.set_zlim(0, max_height * 1.15)

# Keep horizontal dimensions comparable
ax.set_box_aspect((
    max(width, 1),
    max(height_xy, 1),
    max(width, height_xy) * 0.35
))

# Hybrid top-down camera angle
ax.view_init(elev=35, azim=-60)

# Remove axes, gridlines, and pane backgrounds
ax.set_axis_off()

# Remove excess whitespace around the map
plt.subplots_adjust(
    left=0,
    right=1,
    bottom=0,
    top=1
)

plt.show()
```
